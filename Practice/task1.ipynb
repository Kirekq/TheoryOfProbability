### 1.9
Сколько существует семизначных чисел, состоящих из цифр 4, 5, 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза

```python
from itertools import permutations
digits = [4, 4, 4, 5, 5, 6, 6]
a = len(set(permutations(digits)))
a
```

### 1.10
Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) "ТОР"; б) "ТЕОРИЯ"?

```python
from itertools import permutations

item = "ТЕОРИЯ"
k = 3

a = 1 / len(set(permutations(item, r=k)))
a
```

```python
from itertools import permutations

item = "ТЕОРИЯ"
k = 6

a = 1 / len(list(permutations(item, r=k)))
a
```

### 1.16

### 1.17
В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

```python
import math as m
def comb(n, k):
    return m.factorial(n)/(m.factorial(k) * m.factorial(n-k))
ans = comb(9, 2) * comb(8, 7) / comb(36,7)
print(f"{ans:.7f}")
```
